```python
class Node:
    def __init__(self,data):
        self.data = data;
        self.next = None;

class SinglyLinkedList:
    def __init__(self):
        self.head = None;
        self.tail = None;
        self.length = 0;

    def append(self,data):
        newNode = Node(data);
        if not self.head:
            self.head = newNode;
            self.tail = newNode;
        else:
            self.tail.next = newNode;
            self.tail = newNode;
        self.length+=1;
        return self;

    def traverse(self):
        if not self.head:
            return None;
        curr = self.head;
        while(curr != None):
            print(curr.data,end="->");
            curr = curr.next;
        print('None')

    def search(self,ind):
        if(ind <0 or ind > self.length):
            return None;
        cnt = 0;
        curr = self.head;
        while(cnt != ind):
            cnt+=1;
            curr = curr.next;
        return curr.data;

    def middle_of_ll(self):
        if(self.head is None or self.length ==1 ):
            return self.head;
        mid = self.length//2;
        foundNode = self.search(mid)
        return foundNode

```

```python
linkedLL = SinglyLinkedList();
linkedLL.append(10);
linkedLL.append(20);
linkedLL.append(30);
linkedLL.append(40);
```
