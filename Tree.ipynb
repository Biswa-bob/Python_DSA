```python
class TreeNode:
    def __init__(self,val):
        self.val = val;
        self.children = [];

root = TreeNode(1);

child1 = TreeNode(2)
child2 = TreeNode(3)
child3 = TreeNode(4)

root.children.append(child1);
child1.children.append(child2);
root.children.append(child3);

def print_tree(root):
    if(root == None):
        return
    print(root.val);
    for eachChild in root.children:
        print_tree(eachChild)

# print_tree(root)

def print_tree_detail(root):
    if(root == None):
        return
    print(f"{root.val}:",end="");
    for each_child in root.children:
        print(each_child.val,end=",")
    print()
    for eachChild in root.children:
        print_tree_detail(eachChild)

# print_tree_detail(root)

```

### Taking input recursively

```python
def take_input():
    data = int(input("Enter the data for the Node:"))
    node = TreeNode(data);
    num_children = int(input(f"Enter the number of Children for {data}"))

    for _ in range(num_children):
        child = take_input()
        node.children.append(child)

    return node

root = take_input()
print_tree_detail(root);
```

 ### Take Input Level-wise

```python
from collections import deque;

def take_input_levelwise():
    data = int(input("Enter the root data: "))
    root = TreeNode(data);
    queue = deque([root]);
    while(len(queue) !=0):
        curr_node  = queue.popleft();
        num_children = int(input("Enter the number of children for "+ str(curr_node.val)))
        for i in range(num_children):
            child_data = int(input(f"Enter the data for {i+1} child of {curr_node.val}:"))
            child_node = TreeNode(child_data);
            curr_node.children.append(child_node);
            queue.append(child_node)
    return root;

root = take_input_levelwise()
print_tree_detail(root)
```

### Count Nodes of a Tree

```python
def count_nodes(root):
    if(root == None):
        return 0
    number_of_nodes = 1
    for eachChild in root.children:
        print(eachChild.val)
        number_of_nodes = number_of_nodes + count_nodes(eachChild)
    return number_of_nodes


rootCount = count_nodes(root)
print(rootCount)
```

### Height of a Tree

```python
def height_of_tree(root):
    if(root == None):
        return 0
    height = 1
    max_height = 0
    for eachChild in root.children:
       max_height = max(max_height,height_of_tree(eachChild))
    height = height + max_height
    return height

heightOfTree = height_of_tree(root)
print(heightOfTree)
```
