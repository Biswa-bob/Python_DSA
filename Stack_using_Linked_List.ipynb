```python
class Node:
    def __init__(self,val):
        self.val = val;
        self.next = None;

class Stack:
    def __init__(self):
        self.head = None;
        self.size = 0;
    
    def is_empty(self):
        return self.head is None;

    def push(self,val):
        new_node = Node(val);
        new_node.next = self.head;
        self.head = new_node;
        self.size+=1;
    
    def size_stack(self):
        return self.size;

    def pop(self):
        if(self.is_empty()):
            print("Here is the code!")
            return None;
        else:
            print('Here is the pointer')
            pop_node = self.head;
            self.head = self.head.next;
            self.size = self.size -1;
            print(pop_node.val);
            return pop_node.val;

    def peek(self):
        if(self.is_empty()):
            return None;
        print (self.head.val);


```

```python
stack = Stack()
stack.push(10)
stack.push(20)
stack.push(30)
stack.push(40)
stack.push(50)
stack.push(60)
```

```python
stack.size_stack()
```

```python
print(stack.pop())
stack.size_stack()
```
