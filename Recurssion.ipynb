Program to write GFG given number of times

```python
def func1(n):
    while(n<=0):
        return
    print("GFG\n")
    func1(n-1)

func1(5)
```

Program to print Binary representation of a Number

```python
def func(n):
    if n==0:
        return
    print(n%2)
    func(n//2)

func(5)
```

Program to print log(n)

```python
def func(n):
    if n<=1:
        return 0
    else:
        return 1+func(n/2)

func(10)
```
